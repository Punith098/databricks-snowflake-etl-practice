```sql
-- 1️⃣ Create proto table (no Query ID needed)
CREATE OR REPLACE TABLE proto_customers (
    id INT AUTOINCREMENT,
    name STRING,
    email STRING
);
```

```sql
-- 3️⃣ Update data in proto_customers
-- Query ID: Q002
UPDATE proto_customers
SET email = 'alice.new@example.com'
WHERE id = 1;
```

```sql
-- 4️⃣ Delete data from proto_customers
-- Query ID: Q003
DELETE FROM proto_customers
WHERE id = 2;
```

```sql
-- 5️⃣ Create staging table for merge
CREATE OR REPLACE TABLE staging_customers (
    id INT,
    name STRING,
    email STRING
);
```

```sql
-- 6️⃣ Insert data into staging table
-- Query ID: Q004
INSERT INTO staging_customers (id, name, email)
VALUES (1, 'Alice Updated', 'alice.updated@example.com'),
       (3, 'Charlie', 'charlie@example.com');
```

```sql
-- 7️⃣ Merge staging into proto_customers (UPSERT)
-- Query ID: Q005
MERGE INTO proto_customers AS target
USING staging_customers AS source
ON target.id = source.id
WHEN MATCHED THEN
  UPDATE SET target.name = source.name, target.email = source.email
WHEN NOT MATCHED THEN
  INSERT (id, name, email) VALUES (source.id, source.name, source.email);
```
